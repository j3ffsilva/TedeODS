```python
import json
import plotly.graph_objects as go
```

## Cobertura de classificação ODS — Elsevier

```python
with open("data/processed/output_all18_complete.json") as f:
    data = json.load(f)

total = len(data)
elsevier = sum(1 for r in data if any(o in range(1, 17) for o in r.get("ods", [])))
nao_elsevier = total - elsevier

grupos = [
    {"label": "Classificados pela Elsevier",     "n": elsevier,     "pct": round(100 * elsevier / total, 1)},
    {"label": "Não classificados pela Elsevier", "n": nao_elsevier, "pct": round(100 * nao_elsevier / total, 1)},
]

cores = ["#2196F3", "#BDBDBD"]
fonte = "Times New Roman"

fig = go.Figure()

for grupo, cor in zip(grupos, cores):
    fig.add_trace(go.Bar(
        name=grupo["label"],
        x=[grupo["n"]],
        y=[""],
        orientation="h",
        marker_color=cor,
        text=f"{grupo['n']:,} ({grupo['pct']}%)".replace(",", "."),
        textposition="inside",
        insidetextanchor="middle",
        textfont=dict(family=fonte),
        hovertemplate=(
            f"<b>{grupo['label']}</b><br>"
            f"n = {grupo['n']:,}<br>"
            f"{grupo['pct']}%"
            "<extra></extra>"
        ),
    ))

fig.update_layout(
    barmode="stack",
    separators=",.",
    font=dict(family=fonte),
    title=dict(
        text=f"Cobertura de classificação ODS — Elsevier (n = {total:,})".replace(",", "."),
        font=dict(family=fonte, size=15),
    ),
    xaxis=dict(title="Quantidade de teses e dissertações", tickformat=","),
    yaxis=dict(showticklabels=False),
    legend=dict(
        orientation="h",
        yanchor="bottom", y=1.05,
        xanchor="left", x=0,
        font=dict(family=fonte),
    ),
    height=250,
    width=500,
    margin=dict(t=80, b=40, l=20, r=20),
    plot_bgcolor="white",
    paper_bgcolor="white",
)

fig.show()
```

## Distribuição dos trabalhos classificados por ODS (Elsevier)

```python
from collections import Counter

contagem = Counter()
for r in data:
    for o in r.get("ods", []):
        if 1 <= o <= 16:
            contagem[o] += 1

# Ordena do mais ao menos frequente
ods_ordenados = sorted(contagem.items(), key=lambda x: x[1])

labels = [f"ODS {o:02d}" for o, _ in ods_ordenados]
valores = [n for _, n in ods_ordenados]
pcts = [round(100 * n / elsevier, 1) for n in valores]

fonte = "Times New Roman"

fig2 = go.Figure(go.Bar(
    x=valores,
    y=labels,
    orientation="h",
    marker_color="#2196F3",
    text=[f"{v:,} ({p}%)".replace(",", ".") for v, p in zip(valores, pcts)],
    textposition="outside",
    textfont=dict(family=fonte, size=11),
    cliponaxis=False,
    hovertemplate="<b>%{y}</b><br>n = %{x}<extra></extra>",
))

fig2.update_layout(
    separators=",.",
    font=dict(family=fonte),
    title=dict(
        text="Distribuição dos 1.095 trabalhos classificados por ODS — Elsevier",
        font=dict(family=fonte, size=14),
    ),
    xaxis=dict(title="Quantidade de teses e dissertações", tickformat=",", range=[0, 450]),
    yaxis=dict(title=""),
    width=500,
    height=450,
    margin=dict(t=60, b=40, l=70, r=20),
    plot_bgcolor="white",
    paper_bgcolor="white",
)

fig2.show()
```

## Distribuição por ODS — Elsevier + Desigualdade Racial (ODS 18)

```python
contagem2 = Counter()
for r in data:
    for o in r.get("ods", []):
        if 1 <= o <= 16 or o == 18:
            contagem2[o] += 1

ods_ordenados2 = sorted(contagem2.items(), key=lambda x: x[1])

nomes_ods = {18: "ODS 18 — Desigualdade Racial"}
labels2 = [nomes_ods.get(o, f"ODS {o:02d}") for o, _ in ods_ordenados2]
valores2 = [n for _, n in ods_ordenados2]
pcts2 = [round(100 * n / total, 1) for n in valores2]
cores2 = ["#FF7043" if o == 18 else "#2196F3" for o, _ in ods_ordenados2]

fonte = "Times New Roman"

fig3 = go.Figure(go.Bar(
    x=valores2,
    y=labels2,
    orientation="h",
    marker_color=cores2,
    text=[f"{v:,} ({p}%)".replace(",", ".") for v, p in zip(valores2, pcts2)],
    textposition="outside",
    textfont=dict(family=fonte, size=11),
    cliponaxis=False,
    hovertemplate="<b>%{y}</b><br>n = %{x}<extra></extra>",
))

fig3.update_layout(
    separators=",.",
    font=dict(family=fonte),
    title=dict(
        text="Distribuição por ODS — Elsevier e Desigualdade Racial (ODS 18)",
        font=dict(family=fonte, size=14),
    ),
    xaxis=dict(title="Quantidade de teses e dissertações", tickformat=",", range=[0, 450]),
    yaxis=dict(title=""),
    width=500,
    height=500,
    margin=dict(t=60, b=40, l=160, r=20),
    plot_bgcolor="white",
    paper_bgcolor="white",
)

fig3.show()
```
